<h4><b><span style="color:#148F77"> Handling Outliers</span></b></h4>


```python
Q1 = df['total_crimes'].quantile(0.25)
Q3 = df['total_crimes'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df['is_outlier'] = ((df['total_crimes'] < lower) | (df['total_crimes'] > upper))
```

```python
df['total_crimes_log'] = np.log1p(df['total_crimes'])
```

```python
df['is_high_risk'] = df['total_crimes'].apply(lambda x: 1 if x > upper else 0)
```

<h4><b><span style="color:#148F77"> Outlier Handling Strategy</span></b></h4>

- Outliers were identified using the IQR method  
- Since this is crime data, extreme values represent real high-crime regions  
- Therefore, outliers were not removed  
- Log transformation was applied to reduce skewness  
- Outliers were flagged for further risk analysis

<h4><b><span style="color:#148F77"> Question: Which crime type dominates in a selected state?</span></b></h4>


```python
import plotly.express as px
import ipywidgets as widgets

crime_cols = df.columns[6:-3]

state_dropdown = widgets.Dropdown(options=df['state_name'].unique(), description='State')

def plot(state):
    temp = df[df['state_name'] == state][crime_cols].sum().sort_values(ascending=False).head(10).reset_index()
    temp.columns = ['Crime Type', 'Count']
    
    fig = px.bar(temp, x='Count', y='Crime Type', orientation='h',
                 title=f"Crime Distribution in {state}")
    fig.show()

widgets.interact(plot, state=state_dropdown)
```

<h4><b><span style="color:#148F77"> Question: Which states are most dangerous in a selected year?</span></b></h4>



```python
year_slider = widgets.IntSlider(min=2017, max=2022, step=1, value=2022)

def plot(year):
    temp = df[df['year'] == year].groupby('state_name')['total_crimes'].sum().sort_values(ascending=False).head(10).reset_index()

    fig = px.bar(temp, x='total_crimes', y='state_name', orientation='h',
                 title=f"Top States by Crime in {year}")
    fig.show()

widgets.interact(plot, year=year_slider)
```

<h4><b><span style="color:#148F77"> Question: Which states exceed a crime threshold?</span></b></h4>


```python
threshold = widgets.IntSlider(min=1000, max=60000, step=1000, value=20000)

def plot(thresh):
    temp = df.groupby('state_name')['total_crimes'].sum()
    temp = temp[temp > thresh].sort_values(ascending=False).head(10)

    fig = px.bar(temp, x=temp.values, y=temp.index, orientation='h',
                 title=f"States Above Crime Threshold: {thresh}")
    fig.show()

widgets.interact(plot, thresh=threshold)
```

<h4><b><span style="color:#148F77">Question: Which districts drive crime in a selected state?</span></b></h4>


```python
state_dropdown = widgets.Dropdown(options=df['state_name'].unique())

def plot(state):
    temp = df[df['state_name'] == state] \
        .groupby('district_name')['total_crimes'] \
        .sum().sort_values(ascending=False).head(10).reset_index()

    fig = px.bar(temp, x='total_crimes', y='district_name',
                 orientation='h',
                 title=f"Top Districts in {state}")
    fig.show()

widgets.interact(plot, state=state_dropdown)
```

<h4><b><span style="color:#148F77"> Question: How does a specific crime type change over time?</span></b></h4>


```python
crime_cols = df.columns[6:-3]

crime_dropdown = widgets.Dropdown(options=crime_cols, description='Crime')

def plot(crime):
    temp = df.groupby('year')[crime].sum().reset_index()

    fig = px.line(temp, x='year', y=crime, markers=True,
                  title=f"Trend of {crime} Over Years")
    fig.show()

widgets.interact(plot, crime=crime_dropdown)
```
